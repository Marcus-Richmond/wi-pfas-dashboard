```python
import requests

# get url of PFAS open sites
layer_url = 'https://dnrmaps.wi.gov/arcgis2/rest/services/EM_PFAS/EM_PFAS_MAPLAYERS_PUBLIC_EXT/MapServer/1'
```

```python
# define parameters for requests.get function
params = {
    'f': 'json'
}

# store requests.get function
response = requests.get(layer_url, params=params, timeout=30)

# check that .get worked
print(response.status_code)

# return error if http request didn't work
response.raise_for_status()

# store metadata as dict
metadata = response.json()

print(metadata['name'])
print(metadata['geometryType'])
print(metadata['maxRecordCount'])
```

```python
# COUNTS QUERY

# create query URL
query_url = layer_url + '/query'

count_params = {
    'where': '1=1',
    'returnCountOnly': 'true',
    'f': 'json'
}

# store count response
count_response = requests.get(query_url, params=count_params, timeout=30)

# check that response worked
print(count_response.status_code)

# return error if http request didn't work
count_response.raise_for_status()

# store metadata as dict
count_data = count_response.json()

print(count_data)
```

```python
# ATTRIBUTES AND GEOMETRY QUERY

feature_params = {
    'where': '1=1',
    'outFields': '*',
    'returnGeometry': 'true',
    'outSR': 4326,
    'f': 'geojson'
}

# store response
feature_response = requests.get(query_url, params=feature_params, timeout=30)

# check that response worked
print(feature_response.status_code)

# return error if http request didn't work
feature_response.raise_for_status()

# store metadata as dict
feature_data = feature_response.json()

print(feature_data['features'])
```

```python
print(len(feature_data['features']))
print((feature_data['features'][0]))
print((feature_data['features'][0]['properties']['SITENAME']))
print((feature_data['features'][0]['properties']['WTM_X']))
print((feature_data['features'][0]['properties']['WTM_Y']))
```

```python
import geopandas as gpd

# convert response to geodataframe
sites_gdf = gpd.GeoDataFrame.from_features(feature_data['features'], crs='EPSG:4326')

sites_gdf
```
